```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from statsmodels.tsa.arima.model import ARIMA

import plotly.express as px

warnings.filterwarnings("ignore")


df = pd.read_csv(r"C:\Users\knara\Downloads\GlobalWeatherRepository.csv")

print("FIRST 5 ROWS")
display(df.head())

print("\nDATASET SHAPE")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns)

print("\nDATASET INFO")
print(df.info())

print("\nMISSING VALUES")
print(df.isnull().sum())
```

```python

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")

df['last_updated'] = pd.to_datetime(
    df['last_updated']
)

df.drop_duplicates(inplace=True)

df.fillna(method='ffill', inplace=True)

print("\nMISSING VALUES AFTER CLEANING")
print(df.isnull().sum())
```

```python
Q1 = df['temperature_celsius'].quantile(0.25)

Q3 = df['temperature_celsius'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['temperature_celsius'] < lower_bound) |
    (df['temperature_celsius'] > upper_bound)
]

print("\nNUMBER OF OUTLIERS")
print(len(outliers))
```

```python
scaler = StandardScaler()

numerical_columns = [
    'humidity',
    'wind_kph',
    'pressure_mb'
]

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)
```

```python


plt.figure(figsize=(14,6))

plt.plot(
    df['last_updated'],
    df['temperature_celsius']
)

plt.title("Temperature Trend Over Time")

plt.xlabel("Date")

plt.ylabel("Temperature")

plt.show()
```

```python

plt.figure(figsize=(10,5))

sns.histplot(
    df['precip_mm'],
    bins=30
)

plt.title("Precipitation Distribution")

plt.show()
```

```python

plt.figure(figsize=(16,10))

sns.heatmap(
    df.corr(numeric_only=True),
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
nomaly_model = IsolationForest(
    contamination=0.01,
    random_state=42
)

df['anomaly'] = anomaly_model.fit_predict(
    df[['temperature_celsius', 'humidity']]
)

anomalies = df[df['anomaly'] == -1]

print("\nANOMALIES DETECTED")

display(anomalies.head())
```

```python

plt.figure(figsize=(14,6))

plt.plot(
    df['last_updated'],
    df['temperature_celsius'],
    label='Temperature'
)

plt.scatter(
    anomalies['last_updated'],
    anomalies['temperature_celsius'],
    color='red',
    label='Anomaly'
)

plt.legend()

plt.title("Temperature Anomalies")

plt.show()
```

```python


temp_series = df.groupby(
    'last_updated'
)['temperature_celsius'].mean()
```

```python

train_size = int(
    len(temp_series) * 0.8
)

train = temp_series[:train_size]

test = temp_series[train_size:]
```

```python

model = ARIMA(
    train,
    order=(5,1,0)
)

model_fit = model.fit()

forecast = model_fit.forecast(
    steps=len(test)
)
```

```python

mae = mean_absolute_error(
    test,
    forecast
)

rmse = np.sqrt(
    mean_squared_error(test, forecast)
)

r2 = r2_score(
    test,
    forecast
)

print("\nMODEL EVALUATION")

print("MAE :", mae)

print("RMSE :", rmse)

print("R2 SCORE :", r2)
```

```python



plt.figure(figsize=(14,6))

plt.plot(
    test.index,
    test,
    label='Actual'
)

plt.plot(
    test.index,
    forecast,
    label='Forecast'
)

plt.legend()

plt.title("Forecast vs Actual")

plt.show()
```

```python



feature_columns = [
    'humidity',
    'wind_kph',
    'pressure_mb'
]

X = df[feature_columns]

y = df['temperature_celsius']

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X, y)

importance = rf_model.feature_importances_

importance_df = pd.DataFrame({
    'Feature': feature_columns,
    'Importance': importance
})

print("\nFEATURE IMPORTANCE")

display(importance_df)
```

```python


plt.figure(figsize=(8,5))

sns.barplot(
    x='Importance',
    y='Feature',
    data=importance_df
)

plt.title("Feature Importance")

plt.show()
```
